### 6.1 Overall Student Performance

In this section, we calculate basic statistical information about students' final grades (G3).

We examine the average, highest and lowest final grades and count students scoring above 15 and students scoring zero.

```python
print("Total Students:", len(student))

print("Average Final Grade:", student['G3'].mean())

print("Highest Final Grade:", student['G3'].max())

print("Lowest Final Grade:", student['G3'].min())

print("Students Scoring Above 15:", (student['G3'] > 15).sum())

print("Students Scoring 0:", (student['G3'] == 0).sum())
```
